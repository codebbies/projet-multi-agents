```python
import pandas as pd
url = "https://assets.publishing.service.gov.uk/media/6a6c9748862aaf18d9c62ac9/ghg-conversion-factors-2026-flat-format-revised.xlsx"
df = pd.read_excel(url, sheet_name=None)  # charge tous les onglets
print(df.keys())  # voir les noms d'onglets disponibles
```

```python
factors = df['Factors by Category']
print(factors.columns.tolist())
print(factors.head(3))
```

```python
vans = factors[factors.apply(lambda row: row.astype(str).str.contains('van', case=False).any(), axis=1)]
print(vans)
```

```python
raw = pd.read_excel(url, sheet_name='Factors by Category', header=None)
print(raw.head(15).to_string())
```

```python
factors = pd.read_excel(url, sheet_name='Factors by Category', header=5)
factors.columns = ['ID', 'Scope', 'Level1', 'Level2', 'Level3', 'Level4', 'ColumnText', 'UOM', 'GHGUnit', 'Factor2026']

vans_total = factors[
    (factors['Level1'] == 'Delivery vehicles') &
    (factors['GHGUnit'] == 'kg CO2e') &
    (factors['UOM'] == 'km')
]
print(vans_total[['Level2', 'Level3', 'Level4', 'Factor2026']])
```

```python
vans_clean = factors[
    (factors['Level1'] == 'Delivery vehicles') &
    (factors['Level2'] == 'Vans') &
    (factors['GHGUnit'] == 'kg CO2e') &
    (factors['UOM'] == 'km')
].dropna(subset=['Factor2026'])

print(vans_clean[['Level3', 'ColumnText', 'Factor2026']])
```
